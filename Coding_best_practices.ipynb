# **INDENTATION**

1. Use 4 spaces per indentation level.





```python
if condition:
    print("Hello")
```

2. Do not use tabs and spaces together. Prefer spaces.

3. For long lines, you can split the code across multiple lines using (), [], or {}.

```python
result = some_function(
    first_argument,
    second_argument,
    third_argument,
)
```

4. Hanging indentation means the continued lines are clearly separated from the line containing the statement:

```python
def calculate_total(
        price, quantity, discount,
        tax):
    return price * quantity
```

5. Multiline conditions can be formatted like this:

```python
if (age >= 18 and
        has_valid_id):
    print("Allowed")
```

Or, more simply:

```python
if (age >= 18 and
    has_valid_id):
    print("Allowed")
```

6. Closing brackets can either align with the opening statement:

```python
numbers = [
    10,
    20,
    30,
]
```

In simple terms

Python indentation = 4 spaces per block.

```python
def greet(name):
    if name:
        print("Hello", name)
    else:
        print("Hello, stranger")
```

The indentation isn't just for appearance in Python—it defines the structure of the program. Unlike languages that use { } to define blocks, Python uses indentation.

# **COMMENTS IN PYTHON**

According to PEP 8, comments should make code easier to understand and should always be kept updated when the code changes.

1. **General Comments**

         -Comments should normally be complete sentences.

         -The first word should be capitalized.

         -Comments should be clear and easy to understand.

         -Comments that no longer match the code should be updated or removed.

```python
# Calculate the total price.
total = price * quantity
```

2. **Block Comments**

A block comment explains a section of code that follows it.

      -Each line starts with # followed by a space.

      -The comment should have the same indentation level as the code it describes.

      -Separate paragraphs with a line containing just #.

```python
# Calculate the average score of the students.
# This helps us determine the overall performance.

total = sum(scores)
average = total / len(scores)
```

3. **Inline Comments**

An inline comment appears on the same line as the code.

PEP 8 recommends using them sparingly. There should be at least two spaces between the code and #.

```python
x = x + 1  # Compensate for border
```

❌ Avoid obvious comments:

```python
x = x + 1  # Increment x
```

The comment doesn't add useful information because the code already clearly shows what it does.

4. **Documentation Strings (Docstrings)**

A docstring describes what a module, function, class, or method does.

PEP 8 recommends writing docstrings for all public modules, functions, classes, and methods.

## **Multiline docstring**

```python
def calculate_average(scores):
    """Calculate the average score.

    Returns the average of all scores provided.
    """
    return sum(scores) / len(scores)
```

The closing """ should be on a separate line for a multiline docstring.

## **One-line docstring**

For a short description, keep the closing """ on the same line:

```python
def greet(name):
    """Return a greeting message."""
    return f"Hello, {name}!"
```

| Type               | Purpose                                       | Example                        |
| ------------------ | --------------------------------------------- | ------------------------------ |
| **Block comment**  | Explain a section of code                     | `# Calculate the average.`     |
| **Inline comment** | Explain something specific on a line          | `x = x + 1  # Adjust value`    |
| **Docstring**      | Document a function, class, method, or module | `"""Calculate the average."""` |


Easy way to remember:

**Comment** → explains the code to the programmer.

**Docstring** → documents what a Python component does.

# **NAMING CONVENTIONS**

This section of PEP 8 explains how to name variables, functions, classes, modules, constants, and other Python objects consistently.

## 1. Common Naming Styles

| Style                         | Example          | Common use                      |
| ----------------------------- | ---------------- | ------------------------------- |
| `lowercase`                   | `name`           | Variables                       |
| `lower_case_with_underscores` | `student_name`   | Variables & functions           |
| `UPPERCASE`                   | `PI`             | Constants                       |
| `UPPER_CASE_WITH_UNDERSCORES` | `MAX_VALUE`      | Constants                       |
| `CapitalizedWords`            | `StudentDetails` | Classes                         |
| `mixedCase`                   | `studentName`    | Only when existing code uses it |


PEP 8 calls CapitalizedWords CapWords, also commonly called **CamelCase**.

## **2. Underscores Have Special Meanings**



### Single leading underscore   "_name"

Indicates that something is intended for internal use.

```python
_internal_data = 10
```

### Single trailing underscore "name_"

Used when a name conflicts with a Python keyword.

```python
class_ = "Student"
```

Instead of trying to use:

```python
class = "Student"  # Invalid
```

### Double leading underscore "__name"

Used inside classes to trigger name mangling, helping avoid accidental name conflicts with subclasses.

```python
class Student:
    def __init__(self):
        self.__marks = 100
```

### Double leading and trailing __name__

These are special "magic" names provided by Python.

Examples:

```python
__init__
__file__
__import__
```

Don't invent your own names in this form; use documented Python names.

## 3. Names to Avoid

Don't use lowercase l, uppercase O, or uppercase I as single-character variable names because they can easily be confused with 1 and 0.

❌ Avoid:

```python
l = 10
O = 20
I = 30
```

## 4. Module and Package Names

Use short, lowercase names.

```python
student.py
data_analysis.py
```

For packages, lowercase names are preferred, and underscores are discouraged.

5. Class Names

Classes normally use CapWords:

```python
class StudentDetails:
    pass


class DataAnalyzer:
    pass
```

## 6. Function and Variable Names

Use lowercase with underscores.

```python
student_name = "Nandu"

def calculate_average():
    pass
```

PEP 8 recommends the same naming style for both functions and variables.

Easy rule

Functions + Variables → snake_case

Classes → CapWords

Constants → **UPPER_CASE**

## 7. Function Arguments

For instance methods, use self as the first argument:

```python
class Student:
    def display_name(self):
        print("Student")
```

For class methods, use cls:

```python
class Student:
    @classmethod
    def create(cls):
        return cls()
```

## 8. Method and Instance Variable Names

Methods follow the same style as functions:

```python
def calculate_total():
    pass
```

For non-public methods or instance variables, use a single leading underscore:

```python
def _calculate_total():
    pass
```

```python
self._student_marks = 90
```

## 9. Constants

Constants are usually defined at the module level and written in uppercase with underscores.

```python
MAX_OVERFLOW = 100
TOTAL = 500
PI_VALUE = 3.14159
```
